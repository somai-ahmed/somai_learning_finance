<h2 style="text-align:center;">L'équation comptable</h2>

<p style="text-align:center; font-size:20px;">
$$\text{Actifs} = \text{Passifs} + \text{Capitaux propres}$$
</p>

<table style="margin:auto;">
<tr><th>Terme</th><th>Signification</th></tr>
<tr><td><b>Actifs</b></td><td>Ce que l'entreprise <b>possède</b> (trésorerie, ordinateur, stocks...)</td></tr>
<tr><td><b>Passifs</b></td><td>Ce que l'entreprise <b>doit aux autres</b> (prêt bancaire, fournisseurs...)</td></tr>
<tr><td><b>Capitaux propres</b></td><td>La part qui <b>appartient au propriétaire</b> (capital investi + bénéfices)</td></tr>
</table>

<p style="text-align:center;">
Cette équation doit <b>toujours rester équilibrée</b> : chaque opération comptable
affecte au moins deux comptes pour préserver l'équilibre (principe de la partie double).
</p>


<h2 style="text-align:center;">Exemple : création d'entreprise</h2>

<p style="text-align:center;">
1. <b>Avant</b> : le propriétaire a 10 000 (UM) d'argent personnel, l'entreprise a 0 (UM).<br>
2. <b>Opération</b> : il investit 10 000 (UM) en espèces dans l'entreprise.
</p>

<table style="margin:auto;">
<tr><th>Actifs</th><th>Capitaux propres</th></tr>
<tr><td>Trésorerie = 10 000 (UM)</td><td>Capital du propriétaire = 10 000 (UM)</td></tr>
</table>

<p style="text-align:center; font-size:18px;">
$$10\,000 = 0 + 10\,000 \quad \checkmark \text{ l'équilibre est respecté}$$
</p>


````
# Vérification numérique de l'exemple
actifs = 10_000       # Trésorerie
passifs = 0
capitaux_propres = 10_000  # Capital du propriétaire

print(f"Actifs = {actifs}")
print(f"Passifs + Capitaux propres = {passifs + capitaux_propres}")
assert actifs == passifs + capitaux_propres, "Déséquilibre !"
print("✔ L'équilibre est respecté")
````

<h2 style="text-align:center;">Simulation comptable interactive</h2>

<p style="text-align:center;">
On reprend la logique de ton programme (Python/Tkinter et C) : quatre opérations
possibles, chacune affecte deux comptes pour garder l'équation en équilibre.
</p>

<table style="margin:auto;">
<tr><th>#</th><th>Opération</th><th>Actif touché</th><th>Contrepartie</th></tr>
<tr><td>1</td><td>Investir 20 000</td><td>Trésorerie +20 000</td><td>Capital propre +20 000</td></tr>
<tr><td>2</td><td>Acheter ordinateur (3 000)</td><td>Trésorerie −3 000</td><td>Ordinateur +3 000</td></tr>
<tr><td>3</td><td>Emprunt bancaire (10 000)</td><td>Trésorerie +10 000</td><td>Prêt bancaire (passif) +10 000</td></tr>
<tr><td>4</td><td>Vente de services (5 000)</td><td>Trésorerie +5 000</td><td>Revenus (capitaux propres) +5 000</td></tr>
</table>


````
# --- État comptable global ---
cash = 0.0
computer = 0.0
loan = 0.0
owner_capital = 0.0
revenue = 0.0

def refresh():
    assets = cash + computer
    equity = owner_capital + revenue
    print("================ BILAN COMPTABLE ================")
    print(f"Trésorerie       : TND {cash:,.2f}")
    print(f"Ordinateur       : TND {computer:,.2f}")
    print("---------------------------------------------------")
    print(f"Total Actifs     : TND {assets:,.2f}")
    print("---------------------------------------------------")
    print(f"Prêt bancaire    : TND {loan:,.2f}")
    print(f"Capitaux propres : TND {equity:,.2f}")
    print("---------------------------------------------------")
    print(f"Actifs ({assets:,.2f}) = Passifs ({loan:,.2f}) + Capitaux propres ({equity:,.2f})")
    print("===================================================")

refresh()
````

````
def investir():
    global cash, owner_capital
    cash += 20000
    owner_capital += 20000
    refresh()

def acheter_ordinateur():
    global cash, computer
    if cash >= 3000:
        cash -= 3000
        computer += 3000
    else:
        print("Fonds insuffisants pour acheter l'ordinateur.")
    refresh()

def emprunt_bancaire():
    global cash, loan
    cash += 10000
    loan += 10000
    refresh()

def vente_services():
    global cash, revenue
    cash += 5000
    revenue += 5000
    refresh()
````

<h2 style="text-align:center;">Vérifier l'équilibre à tout moment</h2>

````
assets = cash + computer
equity = owner_capital + revenue
print(f"Actifs             = {assets:,.2f}")
print(f"Passifs + Capitaux = {loan + equity:,.2f}")
assert round(assets, 2) == round(loan + equity, 2), "Déséquilibre détecté !"
print("✔ L'équation comptable est respectée")
````

<h2 style="text-align:center;">Version C du simulateur comptable</h2>

<p style="text-align:center;">
Voici l'implementation en langage C de la meme logique comptable, alignee sur
les memes quatre operations et le meme flux <code>refresh()</code> que la version Python
ci-dessus (investir, acheter_ordinateur, emprunt_bancaire, vente_services).
</p>


```c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

/* -----------------------------
   Etat comptable global (equivalent des variables Python)
----------------------------- */
float cash = 0.0f;
float computer = 0.0f;
float loan = 0.0f;
float owner_capital = 0.0f;
float revenue = 0.0f;

/* -----------------------------
   Rafraichissement / affichage (equivalent de refresh())
----------------------------- */
void refresh(void) {
    float assets = cash + computer;
    float equity = owner_capital + revenue;

    printf("================ BILAN COMPTABLE ================\n");
    printf("Tresorerie       : TND %.2f\n", cash);
    printf("Ordinateur       : TND %.2f\n", computer);
    printf("---------------------------------------------------\n");
    printf("Total Actifs     : TND %.2f\n", assets);
    printf("---------------------------------------------------\n");
    printf("Pret bancaire    : TND %.2f\n", loan);
    printf("Capitaux propres : TND %.2f\n", equity);
    printf("---------------------------------------------------\n");
    printf("Actifs (%.2f) = Passifs (%.2f) + Capitaux propres (%.2f)\n",
           assets, loan, equity);
    printf("===================================================\n");
}

/* -----------------------------
   Operations comptables (equivalent des callbacks Python)
----------------------------- */
void investir(void) {
    cash += 20000;
    owner_capital += 20000;
    refresh();
}

void acheter_ordinateur(void) {
    if (cash >= 3000) {
        cash -= 3000;
        computer += 3000;
    } else {
        printf("Fonds insuffisants pour acheter l'ordinateur.\n");
    }
    refresh();
}

void emprunt_bancaire(void) {
    cash += 10000;
    loan += 10000;
    refresh();
}

void vente_services(void) {
    cash += 5000;
    revenue += 5000;
    refresh();
}

/* -----------------------------
   Menu interactif
----------------------------- */
void afficher_menu(void) {
    printf("\n--- Simulation Comptable ---\n");
    printf("1. Investir 20 000  TND\n");
    printf("2. Acheter ordinateur\n");
    printf("3. Emprunt bancaire\n");
    printf("4. Vente de services\n");
    printf("5. Afficher le bilan\n");
    printf("0. Quitter\n");
    printf("Choix : ");
}

int main(void) {
    int choix;
    int continuer = 1;

    refresh();

    while (continuer) {
        afficher_menu();
        if (scanf("%d", &choix) != 1) {
            while (getchar() != '\n');
            continue;
        }

        switch (choix) {
            case 1:
                investir();
                break;
            case 2:
                acheter_ordinateur();
                break;
            case 3:
                emprunt_bancaire();
                break;
            case 4:
                vente_services();
                break;
            case 5:
                refresh();
                break;
            case 0:
                continuer = 0;
                break;
            default:
                printf("Choix invalide.\n");
                break;
        }
    }

    printf("Fin de la simulation.\n");
    return 0;
}
```

# Idée centrale de la comptabilité

Ce notebook explique **l'équation comptable fondamentale** et la met en pratique
avec une petite simulation interactive, comme dans ton diagramme :

```
IDEE CENTRALE DE LA COMPTABILITE
        |
        v
   EQUATION COMPTABLE
        |
        v
ACTIFS = PASSIFS + CAPITAUX PROPRES
```


## L'équation comptable

$$\text{Actifs} = \text{Passifs} + \text{Capitaux propres}$$

| Terme | Signification |
|---|---|
| **Actifs** | Ce que l'entreprise **possède** (trésorerie, ordinateur, stocks...) |
| **Passifs** | Ce que l'entreprise **doit aux autres** (prêt bancaire, fournisseurs...) |
| **Capitaux propres** | La part qui **appartient au propriétaire** (capital investi + bénéfices) |

Cette équation doit **toujours rester équilibrée** : chaque opération comptable
affecte au moins deux comptes pour préserver l'équilibre (principe de la
partie double).


## Exemple : création d'entreprise

1. **Avant** : le propriétaire a 10 000 (UM) d'argent personnel, l'entreprise a 0 (UM).
2. **Opération** : il investit 10 000 (UM) en espèces dans l'entreprise.

Résultat :
- Actifs → Trésorerie = 10 000 (UM)
- Capitaux propres → Capital du propriétaire = 10 000 (UM)

$$10\,000 = 0 + 10\,000 \quad \checkmark \text{ l'équilibre est respecté}$$


````
# Vérification numérique de l'exemple
actifs = 10_000       # Trésorerie
passifs = 0
capitaux_propres = 10_000  # Capital du propriétaire

print(f"Actifs = {actifs}")
print(f"Passifs + Capitaux propres = {passifs + capitaux_propres}")
assert actifs == passifs + capitaux_propres, "Déséquilibre !"
print("✔ L'équilibre est respecté")
````

## Simulation comptable interactive

On reprend la logique de ton programme (Python/Tkinter et C) : quatre opérations
possibles, chacune affecte deux comptes pour garder l'équation en équilibre.

| # | Opération | Actif touché | Contrepartie |
|---|---|---|---|
| 1 | Investir 20 000 | Trésorerie +20 000 | Capital propre +20 000 |
| 2 | Acheter ordinateur (3 000) | Trésorerie −3 000 | Ordinateur +3 000 |
| 3 | Emprunt bancaire (10 000) | Trésorerie +10 000 | Prêt bancaire (passif) +10 000 |
| 4 | Vente de services (5 000) | Trésorerie +5 000 | Revenus (capitaux propres) +5 000 |

Les cellules suivantes réimplémentent ça en Python pur (sans Tkinter, pour rester
dans le notebook), avec un état global et une fonction `refresh()`.


````
# --- État comptable global ---
cash = 0.0
computer = 0.0
loan = 0.0
owner_capital = 0.0
revenue = 0.0

def refresh():
    assets = cash + computer
    equity = owner_capital + revenue
    print("================ BILAN COMPTABLE ================")
    print(f"Trésorerie       : TND {cash:,.2f}")
    print(f"Ordinateur       : TND {computer:,.2f}")
    print("---------------------------------------------------")
    print(f"Total Actifs     : TND {assets:,.2f}")
    print("---------------------------------------------------")
    print(f"Prêt bancaire    : TND {loan:,.2f}")
    print(f"Capitaux propres : TND {equity:,.2f}")
    print("---------------------------------------------------")
    print(f"Actifs ({assets:,.2f}) = Passifs ({loan:,.2f}) + Capitaux propres ({equity:,.2f})")
    print("===================================================")

refresh()
````

````
def investir():
    global cash, owner_capital
    cash += 20000
    owner_capital += 20000
    refresh()

def acheter_ordinateur():
    global cash, computer
    if cash >= 3000:
        cash -= 3000
        computer += 3000
    else:
        print("Fonds insuffisants pour acheter l'ordinateur.")
    refresh()

def emprunt_bancaire():
    global cash, loan
    cash += 10000
    loan += 10000
    refresh()

def vente_services():
    global cash, revenue
    cash += 5000
    revenue += 5000
    refresh()
````

## Vérifier l'équilibre à tout moment


````
assets = cash + computer
equity = owner_capital + revenue
print(f"Actifs             = {assets:,.2f}")
print(f"Passifs + Capitaux = {loan + equity:,.2f}")
assert round(assets, 2) == round(loan + equity, 2), "Déséquilibre détecté !"
print("✔ L'équation comptable est respectée")
````

## Visualiser la composition des actifs / passifs / capitaux propres

(nécessite `matplotlib`, déjà présent dans la plupart des environnements Jupyter)


````
import matplotlib.pyplot as plt

labels = ["Trésorerie", "Ordinateur", "Prêt bancaire", "Capital + Revenus"]
values = [cash, computer, loan, owner_capital + revenue]
colors = ["#4C9AFF", "#83BFFF", "#FF6B6B", "#63E6BE"]

plt.figure(figsize=(6, 4))
plt.bar(labels, values, color=colors)
plt.ylabel("TND")
plt.title("Composition du bilan")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
````
